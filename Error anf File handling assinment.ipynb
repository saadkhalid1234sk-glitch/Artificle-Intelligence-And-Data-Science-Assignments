### Q4

### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
# Q5 solution code here
with open("numbers.txt", "r") as f:
    total = 0
    for line in f:
        total += int (line)
print (f"Total = {total}")
```

### Q6

### Q14 
Ek program banayein jisme `finally` block ho, jo humesha execute ho chahe error aaye ya na aaye.

```python
# Q14 solution code here
try:
    num_1 = int (input("Enter First number: "))
    num_2 = int (input("Enter second number: "))
    result = num_1 / num_2
    print (f"Result : {result}")
    
    
except Exception as e:
    print (f"An error occured {e}")
    
    
    
finally:
    print ("Also Execution Completed.")
```

### Q17
Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
# Q17 solution code here
try:
    marks = int (input("Enter total marks"))
    subjects = int(input("Enter Total subjects"))
    average = marks / subjects
    print (f"Average : {average}")


except ValueError:
    print ("Enter In Exact Numbers: ")
except ZeroDivisionError:
    print ("Not Divisible By Zero: ")
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
# Q18 solution code here
try:
    marks = int (input("Enter total marks"))
    subjects = int(input("Enter Total subjects"))
    average = marks / subjects



except ValueError:
    print ("Enter In Exact Numbers: ")
except ZeroDivisionError:
    print ("Not Divisible By Zero: ")


else:
    print ("Sucessfully Calculated")
    print (f"Average : {average}")
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
# Q19 solution code here
age = int (input("Enter your age: "))
if age <18:
    raise ValueError("Age must be greater than or equal to 18.")
else:
    print ("Age Is Valid: ")
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
# Q20 solution code here
file = None 
try:    
    file = open ("data.txt","r")
    content = file.read()
    print (content)
    file.close()
    
    
except FileNotFoundError:
    print ("File Not Found")
    
finally:
    if file is not None:
        file.close()
        print ("File Close Suceessfully.")
    
    
    
```
