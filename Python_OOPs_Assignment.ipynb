**Q8:** Multiple Inheritance ka istemal karte hue do parent classes `Father` aur `Mother` banayein, aur unse aik child class `Child` derive karein jo dono ke habits inherit kare.

```python
# Write your code here for Q8
# Father Class / Parent Class 
class Father():
    def __init__ (self,hard_working="Hard Working ",punctual= "Punctual",height = "6.4"):
        self.hardworking = hard_working
        self.punctual = punctual
        self.height = height
        
# Mother Class:
class Mother():
    def __init__ (self,kindness="Kind",creativity="Cooking",eye_color="Brown"):
        self.kindness = kindness 
        self.creativity = creativity 
        self.eye_colour = eye_color
        
class Child(Father,Mother):
    def __init__ (self,tech_savvy,hard_working="Hard Working ",punctual= "Punctual",height = "6.4",kindness="Kind",creativity="Cooking",eye_color="Brown"):
        Father.__init__(self,hard_working,punctual,height)
        Mother.__init__(self,kindness,creativity,eye_color)
        self.tech_savvy = tech_savvy
        
        
# Making An Instance From Class
c1 = Child("AI And Machine Learning")

# Output Check
print("Father Trait:", c1.hardworking)
print("Mother Trait:", c1.kindness)
print("Child Trait:", c1.tech_savvy)
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Write your code here for Q10
# Shape Class
class Shape():
    def __init__ (self,color):
        self.color =color

# Circle Class
class Circle(Shape):
    def __init__ (self,color,radius):
        super().__init__ (color)
        self.radius = radius
        
    def show_circle_info (self):
        print (f"""
               =======CIRCLE INFO========
               Shape    :   Circle 
               Color    :   {self.color}
               Radius   :   {self.radius}
               """)
class Square(Shape):
    def __init__ (self,color,side):
        super().__init__ (color)
        self.side   = side
        
    # Method To Show Square Info
    def show_square_info(self):
        print (f"""
               =======SQUARE INFO========
               Shape     :      Square
               Color     :      {self.color}
               Sides     :      {self.side}
               """)
        
        
# Making an instance from class
c1 = Circle("Green",9)
s1 = Square ("Blue",4)
c1.show_circle_info()
s1.show_square_info()
#Hierarchical inheritance is a type of inheritance
#in which multiple child classes inherit from a single
# parent class. In this example, Shape is the parent class,
# while Circle and Square are its child classes. Both inherit
# the color attribute from Shape, while Circle adds radius
# and Square adds side. The super() function calls the 
# parent class constructor.
```

**Q11:** Method Overriding kya hai? Parent class `Bird` banayein jisme `fly()` method ho jo print kare 'Most birds can fly'. Child class `Ostrich` banayein jo is method ko override kare aur print kare 'Ostriches cannot fly'.

```python
# Write your code here for Q11
#Birds Class;
class Bird():
    def fly (self):
        print ("Most birds can fly.")
        
#Ostrich Class;
class Ostrich(Bird):
    # Method Over Riding 
    def fly(self):
        print ("Ostrich Can't fly.")
        
# Making an Instance/Object From Classes;         
b1 = Bird()
b2 = Ostrich()

# Calling The Method
b1.fly()
b2.fly()
#Method overriding is a concept in OOP in which a child class defines a method
#with the same name as a method in the parent class but provides a different implementation.
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
from abc import ABC,abstractmethod
class Payment(ABC):
    @abstractmethod
    def pay(self,amount):
        pass
    
class CreditCardPayment(Payment):

    def pay(self, amount):
        print(f"Paid Rs. {amount} using Credit Card")


class UPIPayment(Payment):

    def pay(self, amount):
        print(f"Paid Rs. {amount} using UPI")
        
        
        
c1 = CreditCardPayment()
c2= UPIPayment()

c1.pay(200)
c2.pay(900)
```
