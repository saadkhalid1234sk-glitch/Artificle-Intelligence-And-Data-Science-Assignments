### Q6: List [2, -5, 10, -8, 0] par loop lagayen. Agar number > 0 hai tou 'Positive', < 0 hai tou 'Negative' aur 0 hai tou 'Zero' print karein.

```python
NUM =[2, -5, 10, -8, 0]
Positive =[]
Negative =[]
Zero =[]
for i in NUM:
    if i >0:
        Positive.append(i)
    elif i < 0:
        Negative.append(i)
    else:
        Zero.append(i)      

print ("Positive",Positive)   
print ("Negative",Negative)     
print ("Zero",Zero)
```

### Q10: Ages ki list [12, 18, 25, 15] lein. Agar age 18 ya us se zyada ho tou 'Adult' print karein, warna 'Minor'.

```python
Adult =[]
Minor=[]

Ages = [12,18,25,15]
for i in Ages:
    if i >= 18:
        Adult.append(i)
    else:
        Minor.append(i) 
print ("Adult",Adult)        
print ("Minor",Minor)        
```

### Q18: 10 se 1 tak reverse loop chalayen. Jab number 5 aaye tou print karein 'Halfway there!'.

```python
for i in reversed(range(1,11)):
    print(i)
    if i == 6:
        print ("Half Way Here",end =" ")
```

### Q19: 1 se 10 tak tamam EVEN numbers ka sum (jama) calculate karein ek variable use kar ke, aur aakhir mein print karein.

```python
total = 0

for i in range(1, 11):
    if i % 2 == 0:
        total += i

print("Sum of even numbers:", total)
```
